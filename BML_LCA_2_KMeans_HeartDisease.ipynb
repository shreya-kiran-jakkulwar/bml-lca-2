```python

# Experiment: Train the system using Heart Disease dataset (UCI ML Repository)
# Determine accuracy using K-Means Clustering

# Step 1: Install and import required libraries
!pip install ucimlrepo

import pandas as pd
import numpy as np
from ucimlrepo import fetch_ucirepo
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score

# Step 2: Load the dataset
heart_disease = fetch_ucirepo(id=45)

x = heart_disease.data.features
y = heart_disease.data.targets

print("Shape of dataset:", x.shape)
x.head()

# Step 3: Data Preprocessing
# target column has values 0-4, converting to binary: 0 = no disease, 1 = disease
y = y.iloc[:, 0]
y = np.where(y > 0, 1, 0)

# filling missing values with median of each column
x = x.fillna(x.median())

# Step 4: Feature Scaling
# K-Means works on distance, so all features need to be on the same scale
sc = StandardScaler()
x_scaled = sc.fit_transform(x)

# Step 5: Applying K-Means Clustering
km = KMeans(n_clusters=2, random_state=42, n_init=10)
y_pred = km.fit_predict(x_scaled)

# Step 6: Matching cluster numbers to actual labels
# KMeans assigns cluster 0/1 randomly, so we map each cluster to its majority actual label
labels = np.zeros(len(y_pred))
for i in range(2):
    mask = (y_pred == i)
    labels[mask] = np.bincount(y[mask]).argmax()

# Step 7: Accuracy of the model
accuracy = accuracy_score(y, labels)
print("Accuracy of K-Means Clustering Model =", accuracy)
print("Accuracy in percentage =", round(accuracy * 100, 2), "%")
```
